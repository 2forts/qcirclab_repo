# Subsection **1.4.2 Creating and visualizing a one-qubit circuit**

```python
from qcirclab import Circuit

# Create a circuit with 1 qubit and 1 classical bit
qc = Circuit(1, 1)

# Apply a Hadamard gate to qubit 0
qc.h(0)

# Measure the qubit into classical bit 0
qc.measure(0, 0)

# Visualize the circuit
print(qc.draw())
```

# Subsection **1.4.3 Simulating the circuit**

```python
# Execute the circuit with repeated shots
result = qc.run(shots=1024, seed=7)

# Retrieve measurement counts
counts = result.counts
print(counts)
```

# Optional: estimated probabilities

```python
print(result.probabilities())
```
